### 从220825-230319的standard catalog里筛选在我的研究区域的事件
纬度范围：26.272301° ~ 27.494801°
经度范围：102.310799° ~ 103.519798°

```python
import pandas as pd

# 读取原始数据文件
df = pd.read_csv(r"F:\BHT\CENCcat-format.txt", names=['year','month','day','hour','minute','second','latitude','longitude','depth','mag_type','magnitude'],
                 delim_whitespace=True)  # 请替换为实际文件名
# 确保经纬度是数值
df['latitude'] = pd.to_numeric(df['latitude'], errors='coerce')
df['longitude'] = pd.to_numeric(df['longitude'], errors='coerce')
# 筛选条件：latitude在26.3~27.4之间，且longitude在102.3~103.5之间
filtered_df = df[
    (df['latitude'].between(26.37, 27.4)) & 
    (df['longitude'].between(102, 103.3))
]

# 保存到新CSV文件
filtered_df.to_csv('f:\BHT\standard_catalog_QJ_alltime.csv', index=False)

print(f"筛选完成，共找到{len(filtered_df)}条符合条件的数据")
print("结果已保存到 filtered_events.csv")
```
